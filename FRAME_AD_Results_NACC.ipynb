## Helper Functions

````python
def compute_ece(y_true, y_proba, n_bins=10):
    bin_boundaries = np.linspace(0, 1, n_bins + 1)
    bin_lowers = bin_boundaries[:-1]
    bin_uppers = bin_boundaries[1:]
    confidences = np.max(y_proba, axis=1)
    predictions = np.argmax(y_proba, axis=1)
    accuracies = (predictions == y_true)
    ece = 0
    for bin_lower, bin_upper in zip(bin_lowers, bin_uppers):
        in_bin = (confidences > bin_lower) & (confidences <= bin_upper)
        prop_in_bin = np.mean(in_bin)
        if prop_in_bin > 0:
            accuracy_in_bin = np.mean(accuracies[in_bin])
            avg_confidence_in_bin = np.mean(confidences[in_bin])
            ece += np.abs(avg_confidence_in_bin - accuracy_in_bin) * prop_in_bin
    return ece

def compute_metrics(y_true, y_pred, y_proba, classes=[0, 1, 2]):
    y_true = np.array(y_true)
    y_bin = label_binarize(y_true, classes=classes)
    acc = accuracy_score(y_true, y_pred)
    bacc = balanced_accuracy_score(y_true, y_pred)
    f1_macro = f1_score(y_true, y_pred, average='macro', zero_division=0)
    mcc = matthews_corrcoef(y_true, y_pred)
    loss = log_loss(y_true, y_proba, labels=classes)
    ece = compute_ece(y_true, y_proba)
    try:
        # Ensure we use average='macro' and multi_class='ovr'
        auc_macro = roc_auc_score(y_true, y_proba, multi_class='ovr', average='macro', labels=classes)
    except:
        auc_macro = np.nan

    brier_avg = np.mean([brier_score_loss(y_bin[:, i], y_proba[:, i]) for i in range(len(classes))])
    cm = confusion_matrix(y_true, y_pred, labels=classes)

    specs = []
    for i in range(len(classes)):
        tn = np.sum(cm) - (np.sum(cm[i, :]) + np.sum(cm[:, i]) - cm[i, i])
        fp = np.sum(cm[:, i]) - cm[i, i]
        specs.append(tn / (tn + fp) if (tn + fp) > 0 else 0)

    res = {
        'Accuracy': acc,
        'Balanced_Acc': bacc,
        'F1_Macro': f1_macro,
        'AUC_Macro': auc_macro,
        'MCC': mcc,
        'LogLoss': loss,
        'Brier': brier_avg,
        'ECE': ece
    }

    p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, labels=classes, zero_division=0)
    for i, cname in enumerate(['CN', 'MCI', 'AD']):
        res[f'{cname}_Prec'] = p[i]
        res[f'{cname}_Rec'] = r[i]
        res[f'{cname}_Spec'] = specs[i]
        res[f'{cname}_F1'] = f[i]

    res['ConfusionMatrix'] = cm
    return res

# Verification test
test_y_true = [0, 1, 2, 0]
test_y_pred = [0, 1, 2, 1]
test_y_proba = np.array([[0.8, 0.1, 0.1], [0.2, 0.7, 0.1], [0.1, 0.2, 0.7], [0.3, 0.6, 0.1]])
test_metrics = compute_metrics(test_y_true, test_y_pred, test_y_proba)
print("Verification - Metrics Keys:", list(test_metrics.keys()))
if 'AUC_Macro' in test_metrics:
    print(" Success: 'AUC_Macro' key found.")
else:
    print(" Error: 'AUC_Macro' key missing.")

def median_impute(X_train, X_test):
    imp = SimpleImputer(strategy='median')
    Xtr_i = pd.DataFrame(imp.fit_transform(X_train), columns=X_train.columns)
    Xte_i = pd.DataFrame(imp.transform(X_test), columns=X_test.columns)
    return Xtr_i, Xte_i

def align_probabilities(raw_proba, model_classes, target_classes=[0, 1, 2]):
    aligned = np.zeros((len(raw_proba), len(target_classes)))
    for i, cls in enumerate(model_classes):
        if cls in target_classes:
            aligned[:, target_classes.index(cls)] = raw_proba[:, i]
    row_sums = aligned.sum(axis=1, keepdims=True)
    row_sums[row_sums == 0] = 1.0
    return aligned / row_sums

def get_tabpfn():
    from tabpfn_client import TabPFNClassifier
    from tabpfn_client.constants import ModelVersion
    return TabPFNClassifier.create_default_for_version(
        ModelVersion.V3_5,
        ignore_pretraining_limits=True
    )
skf = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=RANDOM_SEED)
print("Unified Metrics and Helpers Updated")
````

## Experiment 1: Benchmarking
Compare FRAME-AD (TabPFN 3.5) against 11 methods using stratified 5-fold cross-validation across two groups.

**Group 1 — Median-Imputed Baselines** (missing values filled with feature medians before training):
- Logistic Regression, Random Forest, XGBoost, LightGBM, CatBoost, TabNet, RealMLP

**Group 2 — Native Missing Value Handlers** (no imputation; models handle missingness directly):
- TabPFN v2.5, TabPFN 3.0, TabPFN 3.5 (FRAME-AD), AutoGluon, MITRA

All Group 2 models receive raw data with missing values intact. TabPFN 3.5 is the primary FRAME-AD model. TabPFN v2.5 and 3.0 are included as within-family version comparisons.

````python
import os
import gc
import traceback
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import LabelEncoder, label_binarize
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    roc_auc_score, accuracy_score, f1_score,
    matthews_corrcoef, brier_score_loss, log_loss,
    confusion_matrix, precision_recall_fscore_support
)

warnings.filterwarnings("ignore")

RANDOM_SEED = 42
CV_FOLDS = 5
N_JOBS = -1
OUTPUT_DIR = "results"

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(f"{OUTPUT_DIR}/figures", exist_ok=True)

def save_model_results(model_name, fold_results, filename):
    df_folds = pd.DataFrame(fold_results)
    numeric_df = df_folds.select_dtypes(include=[np.number])

    summary = pd.DataFrame({
        "Metric": numeric_df.columns,
        "Mean": numeric_df.mean().values,
        "Standard Deviation": numeric_df.std(ddof=1).values
    })
    filepath = os.path.join(OUTPUT_DIR, filename)
    summary.to_csv(filepath, index=False)
    return summary

def display_model_execution_summary(model_name, results, X_data):
    df = pd.DataFrame(results)
    print(f"\n{'='*20} {model_name} EVALUATION {'='*20}")

    # Use standardized metric keys
    main_metrics = ['AUC_Macro', 'Accuracy', 'F1_Macro', 'MCC', 'Brier', 'LogLoss']
    available = [m for m in main_metrics if m in df.columns]
    display(df[available].mean().to_frame('Mean Value'))

    fig, ax = plt.subplots(1, 2, figsize=(15, 5))
    agg_cm = np.sum(np.stack(df['ConfusionMatrix'].values), axis=0)
    sns.heatmap(agg_cm, annot=True, fmt='g', cmap='Blues',
                xticklabels=['CN','MCI','AD'], yticklabels=['CN','MCI','AD'], ax=ax[0])
    ax[0].set_title(f'Aggregated Confusion Matrix: {model_name}')

    corr = X_data.corr()
    sns.heatmap(corr, cmap='RdBu_r', center=0, ax=ax[1])
    ax[1].set_title(f'Feature Correlation Matrix')
    plt.tight_layout()
    plt.show()

print("\n Helper functions standardized to use AUC_Macro.")
````

````python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

print("=" * 60)
print("LOGISTIC REGRESSION")
print("=" * 60)
logistic_results = []

for fold, (tr_idx, te_idx) in enumerate(skf.split(X, y), 1):
    Xtr, Xte = X.iloc[tr_idx], X.iloc[te_idx]
    ytr, yte = y.iloc[tr_idx], y.iloc[te_idx]
    Xtr_i, Xte_i = median_impute(Xtr, Xte)

    model = LogisticRegression(max_iter=5000, random_state=RANDOM_SEED, class_weight='balanced')
    model.fit(Xtr_i, ytr)

    pred = model.predict(Xte_i)
    raw_proba = model.predict_proba(Xte_i)
    proba = align_probabilities(raw_proba, model.classes_)

    m = compute_metrics(yte, pred, proba)
    logistic_results.append(m)
    # Changed 'AUC_Macro' to 'AUC_Macro' to match the compute_metrics return key
    print(f"Fold {fold}: AUC={m['AUC_Macro']:.4f}, LogLoss={m['LogLoss']:.4f}")

display_model_execution_summary("Logistic Regression", logistic_results, X)
print("\nClassification Report (Last Fold):")
print(classification_report(yte, pred, target_names=['CN','MCI','AD']))
save_model_results("Logistic Regression", logistic_results, "logistic_regression_results.csv")
````

````python
from sklearn.ensemble import RandomForestClassifier
import gc

print("=" * 60)
print("RANDOM FOREST")
print("=" * 60)

random_forest_results = []

for fold, (tr_idx, te_idx) in enumerate(skf.split(X, y), 1):
    Xtr_f, Xte_f = X.iloc[tr_idx], X.iloc[te_idx]
    ytr_f, yte_f = y.iloc[tr_idx], y.iloc[te_idx]
    Xtr_i, Xte_i = median_impute(Xtr_f, Xte_f)

    model = RandomForestClassifier(n_estimators=300, random_state=RANDOM_SEED, class_weight="balanced", n_jobs=N_JOBS)
    model.fit(Xtr_i, ytr_f)

    pred = model.predict(Xte_i)
    proba = align_probabilities(model.predict_proba(Xte_i), model.classes_)

    m = compute_metrics(yte_f, pred, proba)
    random_forest_results.append(m)
    # Consistently use 'AUC_Macro' to match compute_metrics output
    print(f"Fold {fold}: AUC={m['AUC_Macro']:.4f}, LogLoss={m['LogLoss']:.4f}")

# Display summary including Confusion Matrix
display_model_execution_summary("Random Forest", random_forest_results, X)

# Print Final Classification Report for the last fold
print("\nClassification Report (Last Fold):")
print(classification_report(yte_f, pred, target_names=['CN','MCI','AD']))

save_model_results("Random Forest", random_forest_results, "random_forest_results.csv")
````

````python
# ============================================================
# EXPERIMENT 2: HELPER FUNCTION
# Evaluate one feature subset with 5-fold CV
# Run this cell before Experiment 2 and Experiment 5b
# ============================================================

import os
import gc
import numpy as np
import pandas as pd

metrics_to_collect = [
    "AUC_Macro",
    "Accuracy",
    "Balanced_Acc",
    "F1_Macro",
    "MCC",
    "LogLoss"
]

def evaluate_tabpfn_subset(feature_list):

    Xsub = X[feature_list]
    fold_metrics = []

    for fold_number, (tr_idx, te_idx) in enumerate(
        skf.split(Xsub, y), start=1
    ):
        Xtr = Xsub.iloc[tr_idx].copy()
        Xte = Xsub.iloc[te_idx].copy()
        ytr = y.iloc[tr_idx].copy()
        yte = y.iloc[te_idx].copy()

        clf = get_tabpfn()
        clf.fit(Xtr, ytr)
        raw_proba = clf.predict_proba(Xte)
        proba = align_probabilities(raw_proba, clf.classes_)
        pred = clf.predict(Xte)
        fold_result = compute_metrics(yte, pred, proba)
        fold_metrics.append(fold_result)
        del clf
        gc.collect()

    summary = {}
    for metric in metrics_to_collect:
        values = np.array(
            [result[metric] for result in fold_metrics], dtype=float
        )
        summary[f"{metric}_Mean"] = np.mean(values)
        summary[f"{metric}_SD"] = np.std(values, ddof=1)

    return summary

print("evaluate_tabpfn_subset is ready.")
````

````python
# ============================================================
# EXPERIMENT 2: SEQUENTIAL BACKWARD FEATURE ABLATION
# Start with all 24 features and remove one feature at a time
# ============================================================

print("=" * 70)
print("EXPERIMENT 2: SEQUENTIAL BACKWARD FEATURE ABLATION")
print("=" * 70)

import os
import gc
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

# Start with every feature available in X
current_features = list(X.columns)

# Stop when this many features remain
MIN_FEATURES = 1

# Metric used to decide which feature to remove
# IMPORTANT: must match your compute_metrics() output exactly
SELECTION_METRIC = "AUC_Macro"

# Metrics to save/report
metrics_to_collect = [
    "AUC_Macro",
    "Accuracy",
    "Balanced_Acc",
    "F1_Macro",
    "MCC",
    "LogLoss"
]

# ------------------------------------------------------------
# STORAGE
# ------------------------------------------------------------

e2_selection_history = []
e2_candidate_history = []

print(f"Starting number of features: {len(current_features)}")
print(f"Stopping at: {MIN_FEATURES} features")
print(f"Selection metric: {SELECTION_METRIC}")


# ------------------------------------------------------------
# HELPER FUNCTION
# Evaluate one feature subset with 5-fold CV
# ------------------------------------------------------------

def evaluate_tabpfn_subset(feature_list):

    Xsub = X[feature_list]

    fold_metrics = []

    for fold_number, (tr_idx, te_idx) in enumerate(
        skf.split(Xsub, y),
        start=1
    ):

        Xtr = Xsub.iloc[tr_idx].copy()
        Xte = Xsub.iloc[te_idx].copy()

        ytr = y.iloc[tr_idx].copy()
        yte = y.iloc[te_idx].copy()

        # Fresh TabPFN model for every fold
        clf = get_tabpfn()

        clf.fit(
            Xtr,
            ytr
        )

        raw_proba = clf.predict_proba(
            Xte
        )

        proba = align_probabilities(
            raw_proba,
            clf.classes_
        )

        pred = clf.predict(
            Xte
        )

        fold_result = compute_metrics(
            yte,
            pred,
            proba
        )

        fold_metrics.append(
            fold_result
        )

        del clf
        gc.collect()

    # Calculate mean and SD for each metric
    summary = {}

    for metric in metrics_to_collect:

        values = np.array(
            [result[metric] for result in fold_metrics],
            dtype=float
        )

        summary[f"{metric}_Mean"] = np.mean(values)
        summary[f"{metric}_SD"] = np.std(
            values,
            ddof=1
        )

    return summary


# ============================================================
# STEP 0: EVALUATE FULL 24-FEATURE MODEL
# ============================================================

print("\n" + "-" * 70)
print(f"Evaluating FULL model with {len(current_features)} features")
print("-" * 70)

full_summary = evaluate_tabpfn_subset(
    current_features
)

e2_selection_history.append({
    "Num_Features": len(current_features),
    "Removed_Feature": "None — Full Model",
    "Remaining_Features": " | ".join(current_features),
    **full_summary
})

print(
    f"Full model: "
    f"Macro F1={full_summary['F1_Macro_Mean']:.4f} | "
    f"AUC={full_summary['AUC_Macro_Mean']:.4f} | "
    f"Balanced Acc={full_summary['Balanced_Acc_Mean']:.4f}"
)


# ============================================================
# SEQUENTIAL BACKWARD ELIMINATION
# ============================================================

while len(current_features) > MIN_FEATURES:

    current_n = len(current_features)

    print("\n" + "=" * 70)
    print(
        f"Finding best {current_n - 1}-feature subset "
        f"from current {current_n} features"
    )
    print("=" * 70)

    candidate_results = []

    # --------------------------------------------------------
    # Try removing each currently available feature
    # --------------------------------------------------------

    for candidate_number, feature_to_remove in enumerate(
        current_features,
        start=1
    ):

        candidate_features = [
            feature
            for feature in current_features
            if feature != feature_to_remove
        ]

        print(
            f"[{candidate_number}/{current_n}] "
            f"Testing removal of: {feature_to_remove}"
        )

        try:

            summary = evaluate_tabpfn_subset(
                candidate_features
            )

            candidate_record = {
                "Starting_Num_Features": current_n,
                "Resulting_Num_Features": len(candidate_features),
                "Feature_Removed": feature_to_remove,
                "Remaining_Features": " | ".join(candidate_features),
                **summary
            }

            candidate_results.append(
                candidate_record
            )

            e2_candidate_history.append(
                candidate_record
            )

            print(
                f"    Macro F1="
                f"{summary['F1_Macro_Mean']:.4f} | "
                f"AUC="
                f"{summary['AUC_Macro_Mean']:.4f}"
            )

        except Exception as error:

            print(
                f"    FAILED when removing "
                f"{feature_to_remove}: {error}"
            )

            continue


    # --------------------------------------------------------
    # MAKE SURE AT LEAST ONE CANDIDATE WORKED
    # --------------------------------------------------------

    if len(candidate_results) == 0:

        raise RuntimeError(
            f"No valid candidate subsets were produced "
            f"when starting with {current_n} features."
        )


    # --------------------------------------------------------
    # SELECT BEST CANDIDATE
    # Highest Macro AUC wins
    # --------------------------------------------------------

    best_candidate = max(
        candidate_results,
        key=lambda result:
            result[f"{SELECTION_METRIC}_Mean"]
    )


    best_removed_feature = best_candidate[
        "Feature_Removed"
    ]

    # Permanently remove selected feature
    current_features.remove(
        best_removed_feature
    )


    # --------------------------------------------------------
    # SAVE SELECTED SUBSET
    # --------------------------------------------------------

    selected_record = {
        "Num_Features": len(current_features),
        "Removed_Feature": best_removed_feature,
        "Remaining_Features": " | ".join(current_features)
    }

    for metric in metrics_to_collect:

        selected_record[
            f"{metric}_Mean"
        ] = best_candidate[
            f"{metric}_Mean"
        ]

        selected_record[
            f"{metric}_SD"
        ] = best_candidate[
            f"{metric}_SD"
        ]


    e2_selection_history.append(
        selected_record
    )


    print("\n*** SELECTED REMOVAL ***")

    print(
        f"Removed: {best_removed_feature}"
    )

    print(
        f"Remaining features: {len(current_features)}"
    )

    print(
        f"Macro F1="
        f"{best_candidate['F1_Macro_Mean']:.4f}"
    )

    print(
        f"Macro AUC="
        f"{best_candidate['AUC_Macro_Mean']:.4f}"
    )

    print(
        f"Balanced Accuracy="
        f"{best_candidate['Balanced_Acc_Mean']:.4f}"
    )


    # --------------------------------------------------------
    # SAVE PROGRESS AFTER EVERY ITERATION
    # Important if Colab disconnects
    # --------------------------------------------------------

    progress_df = pd.DataFrame(
        e2_selection_history
    )

    progress_df.to_csv(
        f"{OUTPUT_DIR}/E2_backward_ablation_progress.csv",
        index=False
    )

    candidates_df = pd.DataFrame(
        e2_candidate_history
    )

    candidates_df.to_csv(
        f"{OUTPUT_DIR}/E2_backward_ablation_all_candidates.csv",
        index=False
    )

    print(
        f"Progress saved at "
        f"{len(current_features)} features."
    )


# ============================================================
# FINAL SELECTED PATH
# ============================================================

e2_df = pd.DataFrame(
    e2_selection_history
)

# Sort from 24 features down to 1
e2_df = e2_df.sort_values(
    "Num_Features",
    ascending=False
).reset_index(drop=True)


# ------------------------------------------------------------
# CREATE DISPLAY COLUMNS WITH MEAN ± SD
# ------------------------------------------------------------

for metric in metrics_to_collect:

    e2_df[metric] = e2_df.apply(
        lambda row:
            f"{row[f'{metric}_Mean']:.4f} ± "
            f"{row[f'{metric}_SD']:.4f}",
        axis=1
    )


# ------------------------------------------------------------
# SAVE FINAL RESULTS
# ------------------------------------------------------------

final_file = (
    f"{OUTPUT_DIR}/"
    "E2_sequential_backward_ablation.csv"
)

e2_df.to_csv(
    final_file,
    index=False
)


# ============================================================
# FIND BEST FEATURE SUBSET
# ============================================================

best_row_index = e2_df[
    "AUC_Macro_Mean"
].idxmax()

best_subset = e2_df.loc[
    best_row_index
]

print("\n" + "=" * 70)
print("EXPERIMENT 2 COMPLETE")
print("=" * 70)

print(
    e2_df[
        [
            "Num_Features",
            "Removed_Feature",
            "AUC_Macro",
            "Accuracy",
            "Balanced_Acc",
            "F1_Macro",
            "MCC",
            "LogLoss"
        ]
    ].to_string(index=False)
)


print("\n" + "-" * 70)
print("BEST FEATURE SUBSET")
print("-" * 70)

print(
    "Number of features:",
    int(best_subset["Num_Features"])
)

print(
    "Macro F1:",
    f"{best_subset['F1_Macro_Mean']:.4f}"
)

print(
    "Macro AUC:",
    f"{best_subset['AUC_Macro_Mean']:.4f}"
)

print(
    "\nRemaining features:"
)

for feature in best_subset[
    "Remaining_Features"
].split(" | "):

    print(" -", feature)


print(
    f"\n Final results saved: {final_file}"
)

print(
    f" All candidate subsets saved: "
    f"{OUTPUT_DIR}/E2_backward_ablation_all_candidates.csv"
)

print(
    f" Progress file saved: "
    f"{OUTPUT_DIR}/E2_backward_ablation_progress.csv"
)
````

## Section 3.8: Individual-Level Patient Profiling

**From the Methods:**
> *A distinguishing feature of FRAME-AD is its capacity to generate individual-level patient profiles that extend beyond aggregate performance metrics to produce personalized computational representations of each patient's disease state.*

For each patient **i** with feature vector **xᵢ**, FRAME-AD produces:

> **Profile(i) = { P(CN | xᵢ),  P(MCI | xᵢ),  P(AD | xᵢ),  φᵢ }**

where φᵢ is the individual SHAP attribution vector identifying biomarkers driving that patient's prediction.

This cell generates and visualizes individual patient profiles for a representative sample of test patients.


````python
# ── NOTE ─────────────────────────────────────────────────────
# This cell requires the following variables from earlier cells:
#   clf_shap     — fitted TabPFN model (from E5 SHAP Setup cell)
#   explainer    — fitted KernelExplainer (from E5 SHAP Setup cell)
#   shap_values  — computed SHAP values (from E5 SHAP Setup cell)
#   Xshap        — SHAP test sample (from E5 SHAP Setup cell)
#   yshap        — true labels for SHAP sample
#   FEATURE_LABELS — human-readable feature names
# ─────────────────────────────────────────────────────────────

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

CLASS_NAMES  = {0: 'CN', 1: 'MCI', 2: 'AD'}
CLASS_COLORS = {'CN': '#028090', 'MCI': '#F59E0B', 'AD': '#E11D48'}

# Get predicted probabilities for all SHAP test samples
all_probas = clf_shap.predict_proba(Xshap)   # shape [n_samples, 3]
all_preds  = all_probas.argmax(axis=1)

# ── Build profile dataframe ───────────────────────────────────
profiles = pd.DataFrame({
    'True_Label':    [CLASS_NAMES[y] for y in yshap.values],
    'Pred_Label':    [CLASS_NAMES[p] for p in all_preds],
    'P_CN':          all_probas[:, 0].round(4),
    'P_MCI':         all_probas[:, 1].round(4),
    'P_AD':          all_probas[:, 2].round(4),
    'Correct':       (all_preds == yshap.values),
}, index=Xshap.index)

# Add top biomarker (highest mean |SHAP| across classes)
mean_shap_per_patient = np.mean(
    [np.abs(shap_values[ci]) for ci in range(3)], axis=0
)  # shape [n_samples, n_features]

feat_names    = list(Xshap.rename(columns=FEATURE_LABELS).columns)
top_feat_idx  = mean_shap_per_patient.argmax(axis=1)
profiles['Top_Biomarker'] = [feat_names[i] for i in top_feat_idx]

print(f" Individual profiles generated for {len(profiles)} test patients")
print(f"\nProfile sample (first 10 rows):")
print(profiles.head(10).to_string())
print(f"\nOverall prediction accuracy on SHAP sample: "
      f"{profiles['Correct'].mean()*100:.1f}%")
````

### Step 1 — Generate Individual Patient Profiles

````python
# Select 3 representative patients:
# 1 correctly classified per diagnostic class (CN, MCI, AD)
selected = []
for true_label in ['CN', 'MCI', 'AD']:
    subset = profiles[
        (profiles['True_Label'] == true_label) &
        (profiles['Correct'] == True)
    ]
    if len(subset) >= 1:
        # Sample exactly 1 per class
        selected.extend(subset.sample(1, random_state=42).index.tolist())

print(f"Selected {len(selected)} representative patients for profiling")

# ── Plot: probability bar + SHAP waterfall per patient ────────
# Adjusted figure height for 3 patients
fig = plt.figure(figsize=(20, len(selected) * 4.5))
gs  = gridspec.GridSpec(len(selected), 2, figure=fig,
                        hspace=0.5, wspace=0.35)

for row_idx, pat_idx in enumerate(selected):
    loc       = profiles.index.get_loc(pat_idx)
    true_lbl  = profiles.loc[pat_idx, 'True_Label']
    pred_lbl  = profiles.loc[pat_idx, 'Pred_Label']
    p_cn      = profiles.loc[pat_idx, 'P_CN']
    p_mci     = profiles.loc[pat_idx, 'P_MCI']
    p_ad      = profiles.loc[pat_idx, 'P_AD']
    correct   = profiles.loc[pat_idx, 'Correct']
    status    = '✓ Correct' if correct else '✗ Incorrect'

    # ── LEFT: Probability distribution ──────────────────────
    ax_prob = fig.add_subplot(gs[row_idx, 0])
    classes = ['CN', 'MCI', 'AD']
    probs   = [p_cn, p_mci, p_ad]
    bar_colors = [CLASS_COLORS[c] for c in classes]
    bars = ax_prob.bar(classes, probs, color=bar_colors, alpha=0.85,
                       edgecolor='white', width=0.6)

    # Highlight predicted class
    pred_idx = classes.index(pred_lbl)
    bars[pred_idx].set_edgecolor('black')
    bars[pred_idx].set_linewidth(2)

    # Add probability labels on bars
    for bar, prob in zip(bars, probs):
        ax_prob.text(bar.get_x() + bar.get_width()/2,
                     bar.get_height() + 0.01,
                     f'{prob:.2f}', ha='center', va='bottom',
                     fontsize=9, fontweight='bold')

    ax_prob.set_ylim(0, 1.12)
    ax_prob.set_ylabel('Posterior Probability', fontsize=9)
    ax_prob.set_title(
        f'Patient {row_idx+1}  |  True: {true_lbl}  →  Pred: {pred_lbl}  [{status}]',
        fontsize=10, fontweight='bold',
        color='#028090' if correct else '#E11D48'
    )
    ax_prob.axhline(0.5, color='gray', linestyle='--', alpha=0.4, linewidth=0.8)
    ax_prob.grid(axis='y', alpha=0.2)
    ax_prob.spines['top'].set_visible(False)
    ax_prob.spines['right'].set_visible(False)

    # ── RIGHT: Top SHAP features (waterfall style) ───────────
    ax_shap = fig.add_subplot(gs[row_idx, 1])

    # Average SHAP across classes for this patient
    sv_patient = np.mean(
        [shap_values[ci][loc] for ci in range(3)], axis=0
    )
    feat_names_display = list(Xshap.rename(columns=FEATURE_LABELS).columns)

    # Top 8 features by absolute SHAP
    top_n   = 8
    abs_sv  = np.abs(sv_patient)
    top_idx = np.argsort(abs_sv)[-top_n:]
    sv_top  = sv_patient[top_idx]
    fn_top  = [feat_names_display[i] for i in top_idx]

    bar_cols = ['#E11D48' if v > 0 else '#028090' for v in sv_top]
    ax_shap.barh(fn_top, sv_top, color=bar_cols, alpha=0.82, edgecolor='white')
    ax_shap.axvline(0, color='black', linewidth=0.8)
    ax_shap.set_xlabel('SHAP Value (mean across classes)', fontsize=9)
    ax_shap.set_title(f'Top {top_n} Biomarker Contributions',
                      fontsize=10, fontweight='bold')
    ax_shap.grid(axis='x', alpha=0.2)
    ax_shap.spines['top'].set_visible(False)
    ax_shap.spines['right'].set_visible(False)

plt.suptitle(
    'FRAME-AD: Individual-Level Patient Profiles (3-Patient Subset)\n'
    'Left: Posterior Disease State Probabilities  |  '
    'Right: Biomarker SHAP Contributions',
    fontsize=13, fontweight='bold', y=1.01
)
plt.savefig(f"{OUTPUT_DIR}/figures/S38_individual_profiles_3pt.png",
            dpi=150, bbox_inches='tight')
plt.show()
print(f"\n Figure saved: {OUTPUT_DIR}/figures/S38_individual_profiles_3pt.png")
````
